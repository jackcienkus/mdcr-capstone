## Data Loading

Contractor B Demographic Data

```python
doc = pymupdf.open("data/Contractor B Workforce Data Form Current Year.pdf")
page = doc[0]
table = page.find_tables()
raw_current_year = table[0].extract()
raw_current_year = raw_current_year[7:]

doc = pymupdf.open("data/Contractor B Workforce Data Form Previous Year.pdf")
page = doc[0]
table = page.find_tables()
raw_prev_year = table[0].extract()
raw_prev_year = raw_prev_year[7:]
```

```python
def pdf_data_loader(raw):
    cols = ['EEO Job Categories', 'Hispanic/Latino Male', 'Hispanic/Latina Female']
    for i in range(3, 9):
        cols.append(raw[0][i].strip('\n').replace('\n', ' ') + " Male")
    for i in range(9, 15):
        cols.append(raw[0][i].strip('\n').replace('\n', ' ') + " Female")
    cols.append("Row Total")
    raw[0] = cols

    df = pd.DataFrame(raw)
    df = df.replace('-', 0)
    df = df.replace('', 0)
    df.columns = df.iloc[0]
    df = df.drop(df.index[0])
    exclude_col = 'EEO Job Categories'
    cols_to_convert = df.columns.difference([exclude_col])
    df[cols_to_convert] = df[cols_to_convert].astype(int)

    return df
```

```python
df_current = pdf_data_loader(raw_current_year)
df_prev = pdf_data_loader(raw_prev_year)
```
