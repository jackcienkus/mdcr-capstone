## Data Loading

Contractor B Demographic Data

```python
doc = pymupdf.open("data/Contractor B Workforce Data Form Current Year.pdf")
page = doc[0]
table = page.find_tables()
raw_current_year = table[0].extract()
raw_current_year = raw_current_year[7:]

doc = pymupdf.open("data/Contractor B Workforce Data Form Previous Year.pdf")
page = doc[0]
table = page.find_tables()
raw_prev_year = table[0].extract()
raw_prev_year = raw_prev_year[7:]
```

```python
def pdf_data_loader(raw):
    cols = ['EEO Job Categories', 'Hispanic/Latino Male', 'Hispanic/Latina Female']
    for i in range(3, 9):
        cols.append(raw[0][i].strip('\n').replace('\n', ' ') + " Male")
    for i in range(9, 15):
        cols.append(raw[0][i].strip('\n').replace('\n', ' ') + " Female")
    cols.append("Row Total")
    raw[0] = cols

    df = pd.DataFrame(raw)
    df = df.replace('-', 0)
    df = df.replace('', 0)
    df.columns = df.iloc[0]
    df = df.drop(df.index[0])
    exclude_col = 'EEO Job Categories'
    cols_to_convert = df.columns.difference([exclude_col])
    df[cols_to_convert] = df[cols_to_convert].astype(int)

    return df
```

```python
df_current = pdf_data_loader(raw_current_year)
df_prev = pdf_data_loader(raw_prev_year)
```

```python
# do analysis here

# Find the percentages of demographics in Michigan by total working population in Michigan
mich_ratio = []
# equation: number of specific working population / total working population
for i in mich_estimates:
    a = mich[i].item() / mich['Estimate!!Total, race and ethnicity!!Total, both sexes!!Number'].item() * 100
    b = round(a, 2)
    mich_ratio.append(b)

# label each value for convenience
mich_ratio = np.column_stack((categories, mich_ratio))
mich_ratio
```

```python
df_categories = ['Hispanic/Latino Male', 'Hispanic/Latina Female',
       'White Male', 'Black or African American Male', 'Asian Male',
       'Native Hawaiian or Other Pacific Islander Male',
       'American Indian or Alaskan Native Male', 'Two or more races Male',
       'White Female', 'Black or African American Female', 'Asian Female',
       'Native Hawaiian or Other Pacific Islander Female',
       'American Indian or Alaskan Native Female', 'Two or more races Female',
       'Row Total']
```

```python
df_prev.columns[1]
```

```python
df_prev[df_prev.columns[14]][11]
```

```python
vals = []

for i in df_categories:
    a = df_current[i][11] / df_current["Row Total"][11] * 100
    b = round(a, 2)
    vals.append(b)
    
contractor_B_ratio_current = np.column_stack((df_categories, vals))
contractor_B_ratio_current

vals = []

for i in df_categories:
    a = df_prev[i][11] / df_prev["Row Total"][11] * 100
    b = round(a, 2)
    vals.append(b)

contractor_B_ratio_prev = np.column_stack((df_categories, vals))
contractor_B_ratio_prev
```

```python
contractor_ratio_current_dict = dict(contractor_B_ratio_current)
mich_ratio_dict = dict(mich_ratio)


result = []

for key, val in contractor_ratio_current_dict.items():
    if key in mich_ratio_dict:
        result.append([key, round(float(val), 2) - round(float(mich_ratio_dict[key]),2)])

result
# positive value is over represented, negative value is under represented
```

```python
contractor_ratio_prev_dict = dict(contractor_B_ratio_prev)

diff_in_year = []

for key, val in contractor_ratio_current_dict.items():
    if key in mich_ratio_dict:
        diff_in_year.append([key, round(float(val), 2) - round(float(contractor_ratio_prev_dict[key]),2)])

diff_in_year
# difference in time. Negative is less than previous year, positive is more than previous year
```
