# Analysis

Let's compare the ratio of black male workers to total workers. We'll do this for Contractor C's workforce, the workforce of Michigan, and the workforce of the United States. Let's start with Contractor C's workforce:

```python
# do analysis here
c_total_workers = float(df["Row Total"][11])
c_total_black_male_workers = float(df["Black or African American Male"][11])

c_black_male_ratio = c_total_black_male_workers / c_total_workers

print(f"Contractor C ratio of Black Male workers to total workforce: {c_black_male_ratio}")
```

The ratio we calculated is $0.0333$, which means that roughly $3.33\%$ of the workforce is made up of black men. 

Now, let's calculate the ratio for the state of Michigan and the entire United States, using ACS data from 2018.

```python
mich_total_workers = float(mich.iloc[0,4])
mich_total_black_male_workers = float(mich.iloc[0,44])

usa_total_workers = float(usa.iloc[0,4])
usa_total_black_male_workers = float(usa.iloc[0,44])

mich_black_male_ratio = mich_total_black_male_workers / mich_total_workers
usa_black_male_ratio = usa_total_black_male_workers / usa_total_workers

print(f"Michigan ratio of Black Male workers to total workforce: {mich_black_male_ratio}")
print(f"USA ratio of Black Male workers to total workforce: {usa_black_male_ratio}")
```

Michigan's workforce has a ratio of $5.71\%$, and the United State's workforce has a ratio of $5.52\%$. Compared to Contractor C's workforce, Contractor C has a lower percentage of black men to total workers.
