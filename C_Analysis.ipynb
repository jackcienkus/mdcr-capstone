# Analysis

Let's compare the ratio of black male workers to total workers. We'll do this for Contractor C's workforce, the workforce of Michigan, and the workforce of the United States. Let's start with Contractor C's workforce:

```python
# do analysis here
c_total_workers = float(current_year_df["Row Total"][11])
c_total_black_male_workers = float(current_year_df["Black or African American Male"][11])

c_black_male_ratio = c_total_black_male_workers / c_total_workers

print(f"Contractor C ratio of Black Male workers to total workforce: {c_black_male_ratio}")
```

The ratio we calculated is $0.0333$, which means that roughly $3.33\%$ of the workforce is made up of black men. 

Now, let's calculate the ratio for the state of Michigan and the entire United States, using ACS data from 2018.

```python
mich_total_workers = float(mich.iloc[0,4])
mich_total_black_male_workers = float(mich.iloc[0,44])

usa_total_workers = float(usa.iloc[0,4])
usa_total_black_male_workers = float(usa.iloc[0,44])

mich_black_male_ratio = mich_total_black_male_workers / mich_total_workers
usa_black_male_ratio = usa_total_black_male_workers / usa_total_workers

print(f"Michigan ratio of Black Male workers to total workforce: {mich_black_male_ratio}")
print(f"USA ratio of Black Male workers to total workforce: {usa_black_male_ratio}")
```

Michigan's workforce has a ratio of $5.71\%$, and the United State's workforce has a ratio of $5.52\%$. Compared to Contractor C's workforce, Contractor C has a lower percentage of black men to total workers.

Let's now look at the ratio of hispanic/latina women to the total workforce. First, we'll check out Contractor C.

```python
c_hispanic_latina_total_workers = float(current_year_df["Hispanic/Latina Female"][11])

c_hispanic_latina_ratio = c_hispanic_latina_total_workers / c_total_workers

print(f"Contractor C ratio of Hispanic/Latina Female workers to total workforce: {c_hispanic_latina_ratio}")
```

The ratio of hispanic/latina women to the total workforce at Contractor C is $0.0351$, or $3.51\%$. With that in mind, let's look at the ratios in Michigan and the United States.

```python
mich_hispanic_latina_total_workers = float(mich.iloc[0,24])
usa_hispanic_latina_total_workers = float(usa.iloc[0,24])

mich_hispanic_latina_ratio = mich_hispanic_latina_total_workers / mich_total_workers
usa_hispanic_latina_ratio = usa_hispanic_latina_total_workers / usa_total_workers

print(f"Michigan ratio of Hispanic/Latina Female workers to total workforce: {mich_hispanic_latina_ratio}")
print(f"USA ratio of Hispanic/Latina Female workers to total workforce: {usa_hispanic_latina_ratio}")
```

The ratio of hispanic/latina women in the Michigan workforce is $0.0206$, or $2.06\%$. This is smaller than that of Contractor C. In terms of the United States, the country has a ratio of $0.0745$, or $7.45\%$. This is higher than both Contractor C and Michigan, though this can be explained by the demographic population of Michigan; According to [Jean Kayitsinga's paper](https://jsri.msu.edu/publications/nexo/vol-xxvii/nexo-volume-xxvii-issue-no-1/latino-population-profiles-in-michigan), the population of hispanic/latino people in Michigan is 564,422 (as of 2020). When compared to the total population of Michigan, which is around 10.1 million, the percentage of hispanic/latino workers we had calculated previously comes into perpective.

Let's automate the process for calculating differences in rates between Contractor C's current year data and previous year data.

```python
current_c_total_workers = c_total_workers
previous_c_total_workers = float(previous_year_df["Row Total"][11])
columns = current_year_df.columns
num_columns = len(columns)
job_roles = list(current_year_df['Job Role'])
num_rows = len(job_roles)
percent_diff_dict = dict()

for col in columns:
    if col != 'Job Role' and col != 'Row Total':
        percent_diff_dict[col] = list()
        for i in range(num_rows):
            prev_total_category = float(previous_year_df[col][i+1])
            curr_total_category = float(current_year_df[col][i+1])
            
            prev_total_ratio = prev_total_category / previous_c_total_workers * 100
            curr_total_ratio = curr_total_category / current_c_total_workers * 100

            ratio_diff = round(curr_total_ratio - prev_total_ratio, 2)

            percent_diff_dict[col].append(ratio_diff)

percent_diff = pd.DataFrame(percent_diff_dict)
percent_diff.iloc[:,0]
```

```python
# num_rows = len(percent_diff.iloc[:,0])
fig, ax = plt.subplots(
    4, 3,
    figsize=(18, 3 * 4 + 4),
    sharex=True,
    # sharey=True,
    layout="constrained",
    squeeze=False,
)
plot_x = 0
plot_y = 0
for i in range(num_rows):
    prev_total_category = int(previous_year_df["Row Total"][i+1])
    curr_total_category = int(current_year_df["Row Total"][i+1])
    pop_diff = curr_total_category-prev_total_category
    ax[plot_x, plot_y].bar(percent_diff.columns, percent_diff.iloc[i], color="blue", edgecolor="black")
    ax[plot_x, plot_y].tick_params("x", rotation=45, rotation_mode="xtick")
    ax[plot_x, plot_y].axhline(0, color="black", linewidth=1)
    if pop_diff < 0:
        ax[plot_x, plot_y].set_title(f"{job_roles[i]} (Pop. change: {pop_diff})")
    else:
        ax[plot_x, plot_y].set_title(f"{job_roles[i]} (Pop. change: +{pop_diff})")
    ax[plot_x, plot_y].set_ylim(-0.3,0.3)
    ax[plot_x, plot_y].set_ylabel("Percent Population Change Per Demographic")
    ax[plot_x, plot_y].grid()
    if plot_y < 2:
        plot_y += 1
    else:
        plot_y = 0
        plot_x += 1
```
